```python
%%writefile qa_chatbot_app.py
import streamlit as st
import os
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq

load_dotenv()
gpk = os.getenv("GPK")

prompt = ChatPromptTemplate.from_messages([("system", "You are a helpful assistant."),("user", "Question: {question}") ])
output_parser = StrOutputParser()

def generate_rep(question, model_name="gemma2-9b-it", temperature=0.7, max_tokens=1000):
    llm = ChatGroq(
        model=model_name,
        groq_api_key=gpk,
        temperature=temperature,
        max_tokens=max_tokens
    )
    chain = prompt | llm | output_parser
    return chain.invoke({"question": question})
    
st.title("Q&A Chatbot using Groq + LangChain")

temperature = st.sidebar.slider("Temperature", min_value=0.0, max_value=1.0, value=0.7)
max_tokens = st.sidebar.slider("Max Tokens", min_value=50, max_value=300, value=150)
model_name = st.sidebar.selectbox("Choose Model", ["gemma2-9b-it", "llama-3.1-8b-instant", "llama-3.3-70b-versatile"], index=0)

user_input = st.text_input("You:")

if user_input:
    with st.spinner("Generating response..."):
        try:
            response = generate_rep(
                question=user_input,
                model_name=model_name,
                temperature=temperature,
                max_tokens=max_tokens
            )
            st.markdown("**Assistant:** " + response)
        except Exception as e:
            st.error(f"Error generating response: {e}")
else:
    st.info("Awaiting your question...")
```
