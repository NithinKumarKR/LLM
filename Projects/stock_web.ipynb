```python
import chromadb
client = chromadb.Client()
collection =client.create_collection(name="collection")
```

```python
collection.add(
    documents=['This Document about Delhi','This Document about US'],
    ids=['id1','id2'] )
```

```python
all_docs = collection.get()
all_docs
```

```python
all_docs = collection.get(ids='id1')
all_docs
```

```python
results = collection.query(query_texts=['darshan'],n_results=2)
results
```

```python
collection.delete(ids = 'id2' )
collection.get()
```

```python
collection.add(
    documents=['This Document about ipo','This Document about karnataka bank'],
    ids=['id1','id2'],
    metadatas=[{"url":"https://www.investorgain.com/report/live-ipo-gmp/331/"},
               {"url":"https://karnatakabank.com/"}]    

)
```
