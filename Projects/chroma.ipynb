```python
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader("https://www.liveipo.in/?type=ALL&category=UPCOMING")
page =loader.load().pop().page_content

#print(page)
```

```python
from langchain_core.prompts import PromptTemplate

prompt_op = PromptTemplate.from_template(
    """
    {page_data}

    Task:
    Extract only IPOs that have an explicitly mentioned Premium value greater than 0.
    Ignore IPOs with no Premium or where Premium is "Not explicitly mentioned".
    Sort the extracted IPOs in descending order of Premium.
    Include all the following details for each IPO:
    - Name
    - Open date
    - Close date
    - Allotment date
    - Listing date
    - Price range
    - Lot size
    - Premium
    - Condition (e.g., Neutral, Subscribe, etc.)
    Output should contain only the filtered IPO details without any extra commentary or notes.
    """
)

chain = prompt_op | llm
op = chain.invoke({"page_data": page})

print(op.content)
```
