```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter

loader=TextLoader("Data_injection/speach.txt")

document =loader.load()

text_spliter=CharacterTextSplitter(chunk_size=200,chunk_overlap=30)

docs=text_spliter.split_documents(document)

```

```python
emb=OllamaEmbeddings(model='gemma:2b')
db = FAISS.from_documents(docs,emb)
print(db)
```

```python
query="what is the current date"
docs=db.similarity_search(query)
docs[0].page_content
```

```python
## Retriver

retr=db.as_retriever()
retr.invoke(query)
```

```python
#Similarty_score
db_score=db.similarity_search_with_score(query)
db_score
```

```python
## Saving and loading 
db.save_local('faiss.index')
```
