```python
from langchain_text_splitters import HTMLHeaderTextSplitter

html_string ="""
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Simple Web Page</title>
    <style>
        body {
            font-family: Arial, sans-serif;
            text-align: center;
            margin: 50px;
        }
        h1 {
            color: #333;
        }
        p {
            color: #666;
        }
    </style>
</head>
<body>
    <h1>Welcome to My Web Page</h1>
    <h1>Section 1</h1>
    <h1>Section 2</h1>
    <h1>Section 3</h1>
    <h1>Section 4</h1>
    <h2>Section 5</h2>
    <p>This is a simple HTML page with basic styling.</p>
</body>
</html>
"""

HTML_split_on =[
    ("h1","Header 1"),
    ("h2","Header 2"),
    ("h3","Header 3"),
    ("h4","Header 4"),]

HTML_splitter=HTMLHeaderTextSplitter(HTML_split_on )
HTML_header_splitter=HTML_splitter.split_text(html_string)
HTML_header_splitter
```
